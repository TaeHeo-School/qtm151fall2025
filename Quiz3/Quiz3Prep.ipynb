3. You will recode a numeric variable into a categorical variable (which takes string values). The category will be based on which interval the numeric variable belongs to.

```python
lat_bins = [-90, 0, 90]
# lat_bins = [float("-inf"), 0, float("inf")]
lat_labels = ["South", "North"]
circuits["hemisphere"] = pd.cut(circuits["lat"], 
                                bins = lat_bins,
                                right = True,
                                labels = lat_labels)
```
