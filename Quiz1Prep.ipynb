```python
#Q1
#1. Use an if/elif/else block to test whether a string belongs to one of two lists. 
# Depending on the outcome (belongs to list A, belongs to list B, doesn't belong to either list) print an appropriate message to the screen.
List1=['apple','banana','cherry', 'date', 'fig', 'grape']
List2=['kiwi','mango','orange','papaya','quince','raspberry']
user_input=input("Enter a fruit name: ")
if user_input in List1:
    print(f"{user_input} is in List 1")
elif user_input in List2:
    print(f"{user_input} is in List 2")
else:
    print(f"{user_input} is not in either list")
```

```python
#Q2
#2. You are given two lists of numbers, x_vals and y_vals, and must create a Numpy array where the components are related to those in the lists
# (for example, you might be asked to create a Numpy array with entries).
#Then generate a histogram of the values in the Numpy array, with labeled axes and a title. 
# You will also be asked to change the color of the bars in the histogram.

import numpy as np
import matplotlib.pyplot as plt
x_vals = []
y_vals = []
for i in range(100):
    x_vals.append(np.random.normal(5,1))
    y_vals.append(np.random.normal(3,0.5))
x=np.array(x_vals)
y=np.array(y_vals)
array = x+y
plt.hist(array, color='red', edgecolor='black')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of x + y')
plt.show()
```

```python
#Q3
#3. Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. 
# Each sub-figure will have its own labeled axes and title.
import numpy as np
import matplotlib.pyplot as plt
# Generate random samples
samples1 = np.random.normal(loc=0, scale=1, size=1000)
samples2 = np.random.normal(loc=0, scale=0.5, size=1000)
# Create subplots
fig, subfig_array = plt.subplots(nrows = 1, ncols = 2,figsize = (6,3))

# First Sub-Figure
subfig_array[0].hist(x = samples1)
subfig_array[0].set_title("Normal Distribution\n with mean 0, s.d. 1")
subfig_array[0].set_xlabel("Sample Value")
subfig_array[0].set_ylabel("Frequency")

# Second Figure
subfig_array[1].hist(x = samples2)
subfig_array[1].set_title("Normal Distribution\n with mean 0, s.d. 0.5")
subfig_array[1].set_xlabel("Sample Value")
subfig_array[1].set_ylabel("Frequency")
# Show the plots
plt.tight_layout()
plt.show()
```

```python
#Q4
# 4. Read in a .csv file to create a Pandas DataFrame. 
# Create a scatter plot with one of the variables (or features) corresponding to the x-axis and another corresponding to the y-axis. 
# The plot should have a title and labeled axes.
import pandas as pd
import matplotlib.pyplot as plt
china_emissions = pd.read_csv('data/china_co2_emissions.csv')
plt.scatter(x = china_emissions['year'], y = china_emissions['total_emissions'])
plt.xlabel("(Year))")
plt.ylabel("(Total Emissions)")
plt.title("Relationship between Year and Total Emissions in China")
plt.show()
```

```python
#Q5
#5. Read in a .csv file to create a Pandas DataFrame. Loop over a list of variable/column names, printing the mean of each variable.
import pandas as pd
import matplotlib.pyplot as plt
mean=[]
cardata=pd.read_csv('data/features.csv')
features={'mpg':cardata['mpg'], 'cylinders':cardata['cylinders'],'displacement':cardata['displacement'],'weight':cardata['weight'], 'acceleration':cardata['acceleration']}
for feature in features:
    mean.append(features[feature].mean())
print(mean)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
mean=[]
china=pd.read_csv('data/china_co2_emissions.csv')
india=pd.read_csv('data/india_co2_emissions.csv')
japan=pd.read_csv('data/japan_co2_emissions.csv')
countries=[china, india, japan]
for country in countries:
    mean.append(country['total_emissions'].mean())
print(mean)
```
